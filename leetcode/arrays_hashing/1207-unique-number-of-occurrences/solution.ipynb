## Solution**Approach:** count values, then ensure counts are all unique.**Complexity:**- Time: O(n)- Space: O(n)(Code is in the next cell; expand it to view.)

```python
from __future__ import annotationsfrom collections import Counterfrom typing import Listclass Solution:    def uniqueOccurrences(self, arr: List[int]) -> bool:        counts = Counter(arr)        occ = list(counts.values())        return len(occ) == len(set(occ))
```
