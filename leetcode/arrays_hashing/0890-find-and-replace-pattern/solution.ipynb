## Solution"""Approach: check bijection between word and pattern via two maps.Time: O(n*k)Space: O(1) (bounded alphabet) / O(k)"""from __future__ import annotationsfrom typing import Listclass Solution:    def findAndReplacePattern(self, words: List[str], pattern: str) -> List[str]:        res: list[str] = []        for w in words:            if len(w) != len(pattern):                continue            w2p: dict[str, str] = {}            p2w: dict[str, str] = {}            ok = True            for wc, pc in zip(w, pattern):                if wc in w2p and w2p[wc] != pc:                    ok = False; break                if pc in p2w and p2w[pc] != wc:                    ok = False; break                w2p[wc] = pc                p2w[pc] = wc            if ok:                res.append(w)        return res
