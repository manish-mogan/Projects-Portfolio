# 1051 — Height CheckerLink: https://leetcode.com/problems/height-checker/## SummaryCount how many indices differ between the original heights array and its sorted version.

## Solution**Approach:** compare array to its sorted version and count mismatches.**Complexity:**- Time: O(n log n)- Space: O(n)(Code is in the next cell; expand it to view.)

```python
from __future__ import annotationsfrom typing import Listclass Solution:    def heightChecker(self, heights: List[int]) -> int:        expected = sorted(heights)        return sum(1 for a, b in zip(heights, expected) if a != b)
```
