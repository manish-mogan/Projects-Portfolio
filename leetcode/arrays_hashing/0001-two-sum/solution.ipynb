## Summary
Given an array of integers and a target, return indices of two distinct elements whose sum equals the target.

## Solution

"""Approach: one-pass hash map (value -> index).
Time: O(n)
Space: O(n)
"""

from __future__ import annotations

from typing import List


class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        seen: dict[int, int] = {}
        for i, num in enumerate(nums):
            need = target - num
            if need in seen:
                return [seen[need], i]
            seen[num] = i
        raise ValueError("No two sum solution")
