# 0525 — Contiguous ArrayLink: https://leetcode.com/problems/contiguous-array/## SummaryFind the maximum length of a contiguous subarray with equal numbers of 0s and 1s.

## Solution**Approach:** prefix balance (0->-1, 1->+1) with first index map.**Complexity:**- Time: O(n)- Space: O(n)(Code is in the next cell; expand it to view.)

```python
from __future__ import annotationsfrom typing import Listclass Solution:    def findMaxLength(self, nums: List[int]) -> int:        first: dict[int, int] = {0: -1}        balance = 0        best = 0        for i, x in enumerate(nums):            balance += 1 if x == 1 else -1            if balance in first:                best = max(best, i - first[balance])            else:                first[balance] = i        return best
```
