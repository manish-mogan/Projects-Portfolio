## Solution**Approach:** compare character counts.**Complexity:**- Time: O(n)- Space: O(1) for fixed alphabet / O(n) general(Code is in the next cell; expand it to view.)

```python
from __future__ import annotationsfrom collections import Counterclass Solution:    def isAnagram(self, s: str, t: str) -> bool:        return Counter(s) == Counter(t)
```
