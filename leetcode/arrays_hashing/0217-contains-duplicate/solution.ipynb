## Summary

## Solution

**Approach:** track seen numbers in a set.

**Complexity:**
- Time: O(n)
- Space: O(n)

(Code is in the next cell; expand it to view.)

```python
"""Approach: track seen numbers in a set.
Time: O(n)
Space: O(n)"""

from typing import List


class Solution:
    def containsDuplicate(self, nums: List[int]) -> bool:
        seen = set()
        for x in nums:
            if x in seen:
                return True
            seen.add(x)
        return False
```
