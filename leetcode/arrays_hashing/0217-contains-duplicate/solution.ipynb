## Solution

"""Approach: track seen numbers in a set.
Time: O(n)
Space: O(n)
"""

from __future__ import annotations

from typing import List


class Solution:
    def containsDuplicate(self, nums: List[int]) -> bool:
        seen: set[int] = set()
        for x in nums:
            if x in seen:
                return True
            seen.add(x)
        return False
