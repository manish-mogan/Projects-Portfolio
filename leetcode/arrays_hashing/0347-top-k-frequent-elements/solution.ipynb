## Solution"""Approach: bucket sort by frequency.Time: O(n)Space: O(n)"""from __future__ import annotationsfrom collections import Counterfrom typing import Listclass Solution:    def topKFrequent(self, nums: List[int], k: int) -> List[int]:        freq = Counter(nums)        buckets: list[list[int]] = [[] for _ in range(len(nums) + 1)]        for num, count in freq.items():            buckets[count].append(num)        res: list[int] = []        for count in range(len(buckets) - 1, 0, -1):            for num in buckets[count]:                res.append(num)                if len(res) == k:                    return res        return res
