## Solution"""Approach: store last index of each value.Time: O(n)Space: O(n)"""from __future__ import annotationsfrom typing import Listclass Solution:    def containsNearbyDuplicate(self, nums: List[int], k: int) -> bool:        last: dict[int, int] = {}        for i, x in enumerate(nums):            if x in last and i - last[x] <= k:                return True            last[x] = i        return False
