## Solution

**Approach:** store the last index of each value in a dictionary. When you see a value again, check if the index distance is `<= k`.

**Complexity:**
- Time: O(n)
- Space: O(n)

(Code is in the next cell; expand it to view.)

```python
"""0219 — Contains Duplicate II
Approach: store last index of each value.
Time: O(n)
Space: O(n)"""

from __future__ import annotations

from typing import List


class Solution:
    def containsNearbyDuplicate(self, nums: List[int], k: int) -> bool:
        last: dict[int, int] = {}
        for i, x in enumerate(nums):
            if x in last and i - last[x] <= k:
                return True
            last[x] = i
        return False
```
