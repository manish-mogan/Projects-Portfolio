# 1014 — Best Sightseeing PairLink: https://leetcode.com/problems/best-sightseeing-pair/## SummaryGiven sightseeing values, maximize score of i<j defined by values[i] + values[j] + i - j.

## Solution**Approach:** keep best (values[i] + i) so far; pair score is best + (values[j] - j).**Complexity:**- Time: O(n)- Space: O(1)(Code is in the next cell; expand it to view.)

```python
from __future__ import annotationsfrom typing import Listclass Solution:    def maxScoreSightseeingPair(self, values: List[int]) -> int:        best = values[0] + 0        ans = 0        for j in range(1, len(values)):            ans = max(ans, best + values[j] - j)            best = max(best, values[j] + j)        return ans
```
