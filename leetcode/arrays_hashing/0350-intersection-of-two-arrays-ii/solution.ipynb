## Solution**Approach:** count occurrences with Counter and take min per value.**Complexity:**- Time: O(n+m)- Space: O(min(n,m))(Code is in the next cell; expand it to view.)

```python
from __future__ import annotationsfrom collections import Counterfrom typing import Listclass Solution:    def intersect(self, nums1: List[int], nums2: List[int]) -> List[int]:        c1 = Counter(nums1)        res: list[int] = []        for x in nums2:            if c1[x] > 0:                res.append(x)                c1[x] -= 1        return res
```
