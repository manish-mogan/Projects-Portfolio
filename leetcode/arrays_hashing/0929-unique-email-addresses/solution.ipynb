# 0929 — Unique Email AddressesLink: https://leetcode.com/problems/unique-email-addresses/## SummaryNormalize email addresses by ignoring dots in the local name and ignoring anything after '+'; count unique normalized addresses.

## Solution**Approach:** normalize local part (ignore dots, ignore after +) and keep domain.**Complexity:**- Time: O(n*k)- Space: O(n)(Code is in the next cell; expand it to view.)

```python
from __future__ import annotationsfrom typing import Listclass Solution:    def numUniqueEmails(self, emails: List[str]) -> int:        seen: set[str] = set()        for email in emails:            local, domain = email.split("@", 1)            if "+" in local:                local = local.split("+", 1)[0]            local = local.replace(".", "")            seen.add(local + "@" + domain)        return len(seen)
```
