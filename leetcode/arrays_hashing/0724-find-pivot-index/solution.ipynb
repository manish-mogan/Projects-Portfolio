# 0724 — Find Pivot IndexLink: https://leetcode.com/problems/find-pivot-index/## SummaryFind an index where the sum of elements to the left equals the sum of elements to the right (pivot index).

## Solution**Approach:** track left sum; right sum is total - left - nums[i].**Complexity:**- Time: O(n)- Space: O(1)(Code is in the next cell; expand it to view.)

```python
from __future__ import annotationsfrom typing import Listclass Solution:    def pivotIndex(self, nums: List[int]) -> int:        total = sum(nums)        left = 0        for i, x in enumerate(nums):            if left == total - left - x:                return i            left += x        return -1
```
