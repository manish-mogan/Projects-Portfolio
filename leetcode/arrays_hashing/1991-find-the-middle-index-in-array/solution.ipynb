# 1991 — Find the Middle Index in ArrayLink: https://leetcode.com/problems/find-the-middle-index-in-array/## SummaryFind the leftmost index where the sum of elements to the left equals the sum of elements to the right (middle index).

## Solution**Approach:** same as pivot index (left sum vs right sum).**Complexity:**- Time: O(n)- Space: O(1)(Code is in the next cell; expand it to view.)

```python
from __future__ import annotationsfrom typing import Listclass Solution:    def findMiddleIndex(self, nums: List[int]) -> int:        total = sum(nums)        left = 0        for i, x in enumerate(nums):            if left == total - left - x:                return i            left += x        return -1
```
