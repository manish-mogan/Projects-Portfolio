# 1426 — Counting ElementsLink: https://leetcode.com/problems/counting-elements/## SummaryCount elements x in an array such that x+1 is also present in the array (counting each occurrence of x).

## Solution**Approach:** put elements in a set; count x where x+1 exists.**Complexity:**- Time: O(n)- Space: O(n)(Code is in the next cell; expand it to view.)

```python
from __future__ import annotationsfrom typing import Listclass Solution:    def countElements(self, arr: List[int]) -> int:        s = set(arr)        return sum(1 for x in arr if x + 1 in s)
```
