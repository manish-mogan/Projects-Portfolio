## Solution**Approach:** prefix sum in-place / build output.**Complexity:**- Time: O(n)- Space: O(1) extra if in-place (output excluded)(Code is in the next cell; expand it to view.)

```python
from __future__ import annotationsfrom typing import Listclass Solution:    def runningSum(self, nums: List[int]) -> List[int]:        total = 0        res: list[int] = []        for x in nums:            total += x            res.append(total)        return res
```
