## Solution"""Approach: mark seen indices by negating nums[abs(x)-1].Time: O(n)Space: O(1) extra (ignoring output)"""from __future__ import annotationsfrom typing import Listclass Solution:    def findDisappearedNumbers(self, nums: List[int]) -> List[int]:        for x in nums:            idx = abs(x) - 1            if nums[idx] > 0:                nums[idx] = -nums[idx]        res: list[int] = []        for i, x in enumerate(nums):            if x > 0:                res.append(i + 1)        return res
