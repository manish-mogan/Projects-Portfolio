## Solution"""Approach: Boyer–Moore voting.Time: O(n)Space: O(1)"""from __future__ import annotationsfrom typing import Listclass Solution:    def majorityElement(self, nums: List[int]) -> int:        candidate = None        count = 0        for x in nums:            if count == 0:                candidate = x                count = 1            elif x == candidate:                count += 1            else:                count -= 1        return candidate  # type: ignore[return-value]
